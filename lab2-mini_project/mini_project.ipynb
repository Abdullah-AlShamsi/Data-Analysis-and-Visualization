```python
start = 5 
systematic_sample = customers.iloc[start::15]
systematic_sample["Satisfaction_Score"].mean()
# The mean Satisfaction_Score for the systematic sample is 3.48, which is equal to the mean of the whole data set.
# The mean Satisfaction_Score for the systematic sample and the random sample is equal to the mean of the whole data set, which indicates that both sampling methods are representative of the population.
# The systematic sample is obtained by selecting every 15th customer from the data set, starting from the 5th customer. This method ensures that the sample is evenly distributed across the data set and reduces the risk of bias that may occur in random sampling.
```

```python
customers["Branch"].unique()
```

```python
Nizwa = customers[customers["Branch"] == "Nizwa"]
Seeb = customers[customers["Branch"] == "Seeb"]
Salalah = customers[customers["Branch"] == "Salalah"]
Sohar = customers[customers["Branch"] == "Sohar"]
Muscat = customers[customers["Branch"] == "Muscat City"]
Sur = customers[customers["Branch"] == "Sur"]


sample_Nizwa = Nizwa.sample(n=15, random_state=42)
sample_Seeb = Seeb.sample(n=15, random_state=42)
sample_Salalah = Salalah.sample(n=15, random_state=42)
sample_Sohar = Sohar.sample(n=15, random_state=42)
sample_Muscat = Muscat.sample(n=15, random_state=42)
sample_Sur = Sur.sample(n=15, random_state=42)


stratified_sample = pd.concat([
    sample_Nizwa,
    sample_Seeb,
    sample_Salalah,
    sample_Sohar, 
    sample_Muscat, 
    sample_Sur
])

stratified_sample.head()
```

```python
stratified_sample["Branch"].unique()
```

```python
stratified_sample["Satisfaction_Score"].mean()
# The mean Satisfaction_Score for the stratified sample is 3.48, which is equal to the mean of the whole data set.
```

```python
selected_branches = ["Sohar", "Nizwa"] 
cluster_sample = customers[ 
    customers["Branch"].isin(selected_branches)
]

cluster_sample["Satisfaction_Score"].mean()
# The mean Satisfaction_Score for the cluster sample is 3.48, which is equal to the mean of the whole data set.
# The mean Satisfaction_Score for the cluster sample is 3.680 and the stratified sample is 3.594, which indicates that the stratified sample is more representative of the population than the cluster sample.
```

```python
biased = customers[customers["Waiting_Time_Min"] > 5]
biased["Satisfaction_Score"].mean()
#the mean Satisfaction_Score for the biased sample is 3.423, which is lower than the mean of the whole data set (3.48). 
#Is Python wrong if the estimate is different? Explain where the problem occurs.
# The problem occurs because the biased sample is not representative of the entire population. It only includes customers with long waiting times, which may have different satisfaction levels than the general customer base.
```

```python
results = pd.Series({
    "Population Mean": customers["Satisfaction_Score"].mean(),
    "Random Sample Mean": random_sample["Satisfaction_Score"].mean(),
    "Systematic Sample Mean": systematic_sample["Satisfaction_Score"].mean(),
    "Stratified Sample Mean": stratified_sample["Satisfaction_Score"].mean(),
    "Cluster Sample Mean": cluster_sample["Satisfaction_Score"].mean(),
    "Biased Sample Mean": biased["Satisfaction_Score"].mean()
})

results.values
```

```python
import matplotlib.pyplot as plt
plt.bar(results.index, results.values) 
plt.xlabel("Sampling Method") 
plt.ylabel("Average Satisfaction Score") 
plt.title("Same Population, Different Samples") 
plt.xticks(rotation=30) 
plt.show()
```

```python
#Which sampling method gives the most noticeably different estimate? Explain why.
# The biased sample gives the most noticeably different estimate. This is because it only includes customers with long waiting times, which may have different satisfaction levels than the general customer base, leading to a non-representative sample.

#Does the bar closest to the population automatically prove that the sampling method is always the best? Explain briefly.
# No, the bar closest to the population does not automatically prove that the sampling method is always the best. While it may indicate that the sample is representative of the population in this specific case, it does not guarantee that the method will always yield accurate results in other situations. The effectiveness of a sampling method can vary depending on the characteristics of the population and the specific research question being addressed.
```
